### **Busca HTML da Página**
---
Nesta etapa, realizamos a busca do conteúdo HTML da página inicial do Hacker News (`https://news.ycombinator.com/`). Utilizamos a biblioteca `requests` para fazer a requisição HTTP GET para a URL especificada.

Verificamos se a requisição foi bem sucedida (código de status 200) e, em caso positivo, armazenamos o conteúdo HTML na variável `html_content`. Este conteúdo será utilizado nas etapas subsequentes para extrair as informações das notícias.

```
import requests

url = 'https://news.ycombinator.com/'
response = requests.get(url)

if response.status_code == 200:
    html_content = response.text
    print(html_content)
```

### **Separação das Notícias**
---
Nesta seção, utilizamos a biblioteca `BeautifulSoup` para analisar o conteúdo HTML obtido da página inicial do Hacker News. O objetivo é extrair os elementos que representam as notícias individuais na página. Identificamos que cada notícia está contida dentro de uma tag `<tr>` com a classe `athing submission`.

```
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_content, 'html.parser')

news_html = soup.find_all('tr', {'class': 'athing submission'})
LIMIT_POSTS = 5

count = 0
for news in news_html:
    if count >= LIMIT_POSTS:
        break

    print(news)

    count+= 1
```

### **Extração e Armazenamento dos Dados das Notícias**
---
Nesta seção, refinamos a extração das informações de cada notícia utilizando a `BeautifulSoup`. Para cada elemento de notícia (`<tr>` com classe `athing submission`) encontrado na etapa anterior, buscamos o ranking (`span` com classe `rank`), o título e a URL (`a` dentro de `span` com classe `titleline`).

As informações extraídas (título e URL) são então armazenadas em um dicionário chamado `posts`, onde a chave é o ranking da notícia. O loop é limitado pelo valor de `LIMIT_POSTS` para processar apenas as primeiras notícias.

```
news_data = {}

count = 0
for news in news_html:
    if count >= LIMIT_POSTS:
        break

    rank = news.find('span', {'class': 'rank'}).text.strip()
    titleHTML = news.find('span', {'class': 'titleline'})

    title = titleHTML.find('a').text.strip()
    url = titleHTML.find('a')['href']

    news_data[rank] = {
        'title': title,
        'url': url
    }

    count+= 1

print("Notícias extraídas:")
for rank, news in news_data.items():
    print(f"  {rank} Título: {news['title']}")
    print(f"    URL: {news['url']}")
```

### **Extração do Conteúdo Principal dos Artigos com `newspaper3k`**
* * *
Nesta seção, para cada notícia extraída, utilizamos a biblioteca `newspaper3k` para acessar a URL do artigo e extrair o conteúdo principal. A biblioteca lida com a complexidade de analisar diferentes estruturas de páginas web para obter o texto relevante do artigo.

```
from newspaper import Article, Config

# Configuração para simular um navegador e evitar bloqueios simples
user_agent = 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36' # Um user-agent real
config = Config()
config.browser_user_agent = user_agent
config.request_timeout = 10

news_data_with_content = []
base_url = 'https://news.ycombinator.com/'

for rank, news in news_data.items():
    print(f"Processando: {news['title']}")
    url = news['url']
    # Construir a URL completa se for um caminho relativo
    if not url.startswith('http'):
        url = base_url + url

    news = Article(url, config=config)
    try:
        news.download()
        news.parse()

        if news.text and len(news.text) > 300: # Um filtro mínimo de sanidade
            news_data_with_content.append({
                'title': news['title'],
                'url': url,
                'original_text': news.text
            })
        else:
            print(f"  -> Falha (pouco texto): {url}")

    except Exception as e:
        print(f"  -> Falha (Download/Parse): {url} | Erro: {e}")

print(f"\nSucesso! {len(news_data_with_content)} de {len(news_data)} post extraídos.")
```

### **Chain de Sumarização**
* * *
Definimos a cadeia de sumarização utilizando o LangChain com o LLM configurado anteriormente. Utilizaremos a técnica de "map_reduce", onde o texto é dividido em partes menores para sumarização individual ("map"), e depois os resumos parciais são combinados em um resumo final ("reduce").

```
from langchain.chains.summarize import load_summarize_chain
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_core.prompts import PromptTemplate

text_splitter = RecursiveCharacterTextSplitter(chunk_size=10000, chunk_overlap=500)

map_prompt = """
INSTRUÇÃO: Você receberá um trecho de um artigo de tecnologia. Sua tarefa é extrair os pontos-chave, fatos principais e argumentos centrais deste trecho.
Seja conciso, factual e denso em informação. O resultado deve ser uma lista de pontos ou um parágrafo curto.
Este é um resumo intermediário, não o produto final.

TRECHO DO TEXTO:
"{text}"

RESUMO INTERMEDIÁRIO (EM INGLÊS):
"""
map_prompt_template = PromptTemplate(template=map_prompt, input_variables=["text"])

combine_prompt = """
CONTEXTO: Você é um editor-chefe sênior em um portal de notícias de tecnologia de ritmo acelerado, como o Hacker News.
TAREFA: Você recebeu vários resumos parciais (pontos-chave) de um artigo. Sua tarefa é sintetizar esses pontos em uma nota de resumo final, coesa e impactante.

PONTOS-CHAVE RECEBIDOS:
"{text}"

REGRAS E RESTRIÇÕES PARA A NOTA FINAL:
1.  **Persona:** Aja como um jornalista sênior.
2.  **Tom:** O tom deve ser jornalístico, direto, informativo e envolvente.
3.  **Formato:** A saída deve ser um (1) único parágrafo.
4.  **Tamanho:** O parágrafo não deve, em hipótese alguma, exceder cinco (5) linhas.
5.  **Idioma:** Escreva a nota final em Inglês (ela será traduzida depois).

NOTA FINAL (1 parágrafo, máx. 5 linhas, em Inglês):
"""
combine_prompt_template = PromptTemplate(template=combine_prompt, input_variables=["text"])

summarize_chain = load_summarize_chain(
    llm,
    chain_type="map_reduce",
    map_prompt=map_prompt_template,
    combine_prompt=combine_prompt_template,
    verbose=True
)
```

### **Chain de Tradução**
* * *
Definimos a cadeia de tradução utilizando o LangChain. Esta cadeia receberá o resumo em inglês gerado pela cadeia de sumarização e o traduzirá para o Português do Brasil (pt-BR).

```
from langchain_core.prompts import PromptTemplate

translate_prompt_template = PromptTemplate(
    input_variables=["english_summary"],
    template="Traduza o seguinte resumo para o Português do Brasil (pt-BR):\n\n{english_summary}\n\nTRADUÇÃO:",
)

translate_chain = translate_prompt_template | llm
```

```
final_news = []

for news in news_data_with_content:
    print(f"\n--- Processando: {news['title']} ---")

    # 1. Preparar os 'Documentos' para o chain de sumarização
    # O TextSplitter cria os chunks
    docs = text_splitter.create_documents([news['original_text']])

    # 2. Executar Sumarização (Técnica 1)
    try:
        summary_output = summarize_chain.invoke(docs)
        english_summary = summary_output['output_text']
        print(f"Resumo (EN): {english_summary[:150]}...")

        # 3. Executar Tradução (Técnica 2)
        translate_output = translate_chain.invoke({"english_summary": english_summary})

        portuguese_summary = translate_output.content
        print(f"Resumo (PT-BR): {portuguese_summary[:150]}...")

        # 4. Armazenar resultados
        final_news.append({
            'title': news['title'],
            'url': news['url'],
            'english_summary': english_summary,
            'portuguese_summary': portuguese_summary
        })

    except Exception as e:
        print(f"  -> Falha no processamento de PLN: {news['title']} | Erro: {e}")
```

#### **Parametrização da Chain do Roteiro de Podcast**
* * *
Definimos o `PromptTemplate` que instruirá o modelo a formatar os resumos das notícias em um roteiro de podcast coeso, incluindo introdução, blocos de notícias e encerramento, tudo em Português do Brasil.

```
from langchain_core.prompts import PromptTemplate

podcast_script_template = """
CONTEXTO: Você é um roteirista de podcast de notícias de tecnologia, criando um script para dois apresentadores: um homem e uma mulher.
TAREFA: Crie um roteiro de podcast envolvente e informativo com base nos resumos de notícias fornecidos. O roteiro deve alternar entre a fala dos dois apresentadores.

INSTRUÇÕES PARA O ROTEIRO:
1.  **Idioma:** O roteiro deve ser inteiramente em Português do Brasil.
2.  **Apresentadores:** Crie um nome para o apresentador masculino e um nome para a apresentadora feminina. Use estes nomes para sinalizar a fala de cada um.
3.  **Estrutura:** O roteiro deve seguir a seguinte estrutura **EXATAMENTE na ordem e formato especificados, começando IMEDIATAMENTE com a Introdução**:
    *   **Introdução:** Uma breve saudação de boas-vindas aos ouvintes e menção rápida de que as notícias de tecnologia serão apresentadas. **Não inclua a apresentação nominal dos apresentadores nesta seção.** Divida a fala desta saudação entre os dois apresentadores.
    *   **Bloco de Notícias:** Para cada notícia fornecida, inclua:
        *   Uma transição suave para a notícia, que pode ser dita por um dos apresentadores.
        *   Apresentação clara do Título da Notícia.
        *   Leitura do Resumo da Notícia (o resumo já está em português). Divida a leitura do resumo ou a apresentação entre os dois apresentadores, alternando a fala.
        *   Uma breve frase de encerramento para a notícia antes de passar para a próxima.
    *   **Encerramento:** Uma breve despedida e agradecimento aos ouvintes, dividida entre os dois apresentadores.
4.  **Formato:** Apresente o roteiro de forma clara, contento APENAS as falas, sem a sinalização dos blocos da estrutura indicando onde é a introdução, bloco de notícias, etc..., e usando os nomes dos apresentadores antes de cada fala, por exemplo:
    NomeHomem: [Fala do homem]
    NomeMulher: [Fala da mulher]
    **Não inclua frases de metalinguagem sobre o roteiro ou instruções para si mesmo.**
5.  **Tom:** O tom deve ser informativo, claro, amigável e envolvente, como em um podcast de notícias popular.
6.  **Conteúdo:** Utilize **apenas** o Título e o Resumo fornecidos para cada notícia. Não adicione informações externas.

RESUMOS DAS NOTÍCIAS:
{news_summaries}

ROTEIRO DO PODCAST (em Português do Brasil, com fala alternada entre dois apresentadores - homem e mulher, com nomes gerados e sinalizados):
"""

podcast_script_prompt = PromptTemplate(
    template=podcast_script_template,
    input_variables=["news_summaries"]
)
```

```
podcast_chain = podcast_script_prompt | llm
```

```
import re

# Remove as marcações de seção usando regex
# As marcações são: **Introdução**, **Bloco de Notícias**, **Notícia seguido por um número**, **Encerramento**
cleaned_podcast_script = re.sub(r'\*\*(Introdução|Bloco de Notícias|Notícia \d+|Encerramento)\*\*\n*', '', podcast_script)

# Opcional: Remover a linha "Aqui está o roteiro do podcast:" se ela existir
cleaned_podcast_script = re.sub(r'Aqui está o roteiro do podcast:\n\n', '', cleaned_podcast_script)


print("--- Roteiro do Podcast Limpo ---")
print(cleaned_podcast_script)
```

```
from google import genai
from google.genai import types
import wave

# Set up the wave file to save the output:
def wave_file(filename, pcm, channels=1, rate=24000, sample_width=2):
   with wave.open(filename, "wb") as wf:
      wf.setnchannels(channels)
      wf.setsampwidth(sample_width)
      wf.setframerate(rate)
      wf.writeframes(pcm)

client = genai.Client( api_key= os.environ['GOOGLE_API_KEY'])
```

```
reponse = client.models.generate_content(
   model = "gemini-2.5-flash-preview-tts",
   contents = cleaned_podcast_script,
   config = types.GenerateContentConfig(
      response_modalities = ["AUDIO"],
      speech_config = types.SpeechConfig(
         multi_speaker_voice_config = types.MultiSpeakerVoiceConfig(
            speaker_voice_configs=[
               types.SpeakerVoiceConfig(
                  speaker='Joao',
                  voice_config=types.VoiceConfig(
                     prebuilt_voice_config=types.PrebuiltVoiceConfig(
                        voice_name='Kore',
                     )
                  )
               ),
               types.SpeakerVoiceConfig(
                  speaker='Maria',
                  voice_config=types.VoiceConfig(
                     prebuilt_voice_config=types.PrebuiltVoiceConfig(
                        voice_name='Puck',
                     )
                  )
               ),
            ]
         )
      )
   )
)

data = reponse.candidates[0].content.parts[0].inline_data.data

file_name='podcast.wav'
wave_file(file_name, data)
```

```
import contextlib
import wave
from IPython.display import Audio

file_index = 0

@contextlib.contextmanager
def wave_file(filename, channels=1, rate=24000, sample_width=2):
    with wave.open(filename, "wb") as wf:
        wf.setnchannels(channels)
        wf.setsampwidth(sample_width)
        wf.setframerate(rate)
        yield wf

def play_audio_blob(blob):
  global file_index
  file_index += 1

  fname = f'audio_{file_index}.wav'
  with wave_file(fname) as wav:
    wav.writeframes(blob.data)

  return Audio(fname, autoplay=True)

def play_audio(response):
    return play_audio_blob(response.candidates[0].content.parts[0].inline_data)
```
